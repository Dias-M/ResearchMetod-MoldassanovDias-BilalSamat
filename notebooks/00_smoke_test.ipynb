# 00 Smoke test
Checks that the environment, package import and the metric code work. Uses synthetic data only.

```python
import sys; sys.path.insert(0, '../src')
import pandas as pd
from phishrobust.metrics import threshold_at_fpr, evaluate
from phishrobust.models import TfidfLR
df = pd.read_csv('../data/sample/emails.csv')
print(df.shape, df.split.value_counts().to_dict())
```

```python
tr, va, te = (df[df.split == s] for s in ('train', 'val', 'test'))
m = TfidfLR().fit(tr.text, tr.label)
thr = threshold_at_fpr(m.score(va[va.label == 0].text), 0.01)
evaluate(te.label, m.score(te.text), thr)
```
